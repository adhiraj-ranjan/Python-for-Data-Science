### Problem Statement

Given a matrix mat of size 3 x 3. Find the maximum numbers from each row, say N1, N2 and N3. Result in a matrix by adding:

```python
 
import numpy as np
mat = np.array([[10,5,9],
                [2,20,6],
                [8,3,30]]).reshape(3,3)
 
```

```python
sum = mat.sum(axis=1)
N1 = sum[0]
N2 = sum[1]
N3 = sum[2]
print(N1, N2, N3)
```
