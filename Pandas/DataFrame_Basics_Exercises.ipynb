````python
# given dataframe

import pandas as pd

df = pd.DataFrame(
    [[0.23, 'f1'], [5.36, 'f2']],
    index=list('pq'),
    columns=list('ab')
)
print(df)
````

````python

# changing column name from a to A
df.rename(columns = {'a':'A'}, inplace = True) 
df
````

````python
# adding a new column c filled with random values
import numpy as np
df['C'] = np.random.rand(2)
````

````python
df
````

````python
# changing the values of column A to complex type
df['A'] = df['A'].astype(complex)
````

````python
df
````

````python
# Displaying rows whose any of the element matches with any element of the given list
lst = ['f30','f50','f2','f0']

print(df[df.isin(lst).any(axis=1)])
````
