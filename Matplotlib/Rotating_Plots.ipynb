### Rotating Plots Tryout

```python
# 3D Scatter Plot

import numpy as np
import matplotlib.pyplot as plt

x, y, z = np.random.rand(3,50)

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

ax.scatter(x, y, z)

ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')

plt.show() 
```

```python
from mpl_toolkits.mplot3d import axes3d
```

```python
for angle in range(0, 360):
    ax.view_init(30, angle)
    plt.draw()
    plt.pause(.001)
     
```
