````python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(0.0, 5.0, 0.1)
y = np.cos(2 * np.pi * x) * np.exp(-x)

plt.plot(x, y, 'o-', markersize=4, label='Damped cosine')

plt.title("Main heading")
plt.suptitle("Figure Title", y=0.94)

plt.xlabel("Abscissa")
plt.ylabel("Ordinate")

plt.annotate(
    r'$\theta = 60^\circ$',
    xy=(1.0, 0.35),
    xytext=(1.0, 0.4)
)

plt.annotate(
    '2nd crest',
    xy=(2.0, 0.14),
    xytext=(2.4, 0.28),
    arrowprops=dict(facecolor='green', shrink=0.05)
)

plt.grid(True)
plt.show()
````
