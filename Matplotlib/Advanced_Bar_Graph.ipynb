### Advanced Bar Graph

Previously, we have seen how to plot a simple bar graph with some given groups. Now, let us improve our graph by adding some details like labels, legend, and colors.

Adding distinct colors

We are aware of the number of groups i.e. 18 therefore, we will create some 18 random colors for each of our group.


```python
           
import random
import numpy as np
# Creating 18 random colors range (0 - 1)
clrs = np.linspace( 0, 1, 18 )
random.shuffle(clrs)
# Creating final list of 18 random colors
colors = []
for i in range(0, 72, 4):
    idx = np.arange( 0, 18, 1 )
    random.shuffle(idx)
    r = clrs[idx[0]]
    g = clrs[idx[1]]
    b = clrs[idx[2]]
    a = clrs[idx[3]]
    colors.append([r, g, b, a])
     
```

```python
colors
```


Next, we pass these random colors into the bar graph. Also, we are not providing label argument within the bar function which will be taken care of inside legend function.

```python
import matplotlib.pyplot as plt          
bar_graph = plt.bar(type_1_grps, type_1_count, bar_width,
                 alpha = 0.5,   # tranparency factor
                 color = colors)   # color factor
                  
```

### Legend

We saw only one label inside the legend section in our last plot. This time we're setting up labels particular to each group name. This can be done as follows:

```python
           
plt.legend(bar_graph, 
           type_1_names,                    # List of group names
           bbox_to_anchor=(1.128, 1.015))   # Position of legend
            
```

### Adding text to plot

Next, we're going to add names of X-axis and Y-axis using xlabel() and ylabel() functions. We can also add a suitable title using title() function along with defining the visibility of grid and setting limit on axis.

```python
           
plt.xticks(type_1_grps + bar_width/2, type_1_names)
plt.xlabel('Type')
plt.ylabel('Pokemon count')
plt.title('Number of Pokemon per Type_1')
plt.grid()
plt.ylim(0,130)
 
```
