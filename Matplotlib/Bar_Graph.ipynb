### plotting bar graph from data

```python
# reading a csv file

import pandas as pd

df = pd.read_csv('pokemon_alopez247.csv')
df.head()
```

let us count the number of distinct groups along with the number of Pokemon(s) belonging to that group available under Type_1 category. We use scipy.stats module to import itemfreq function which results in the total number of distinct samples with their frequency.

```python
import numpy as np

# print(df.iloc[:,2])
type_1 = np.unique(df.iloc[:,2], return_counts=True)
type_1 = np.column_stack(type_1)

print(type_1)

type_1_grps = len(type_1)
print(type_1_grps)

type_1_names = type_1[:,0]
print(type_1_names)

type_1_count = type_1[:,1]
print(type_1_count)
```

```python
import matplotlib.pyplot as plt
           
type_1_grps = np.arange(len(type_1_names))
bar_width = 0.5
plt.bar(type_1_grps, type_1_count, bar_width,
                 alpha = 0.5,   # tranparency factor
                 color = 'g',   # color factor
                 label='Pokemon count respective to their Type_1')
plt.legend(loc='best')
plt.xticks(type_1_grps + bar_width/2, type_1_names)
 
```
